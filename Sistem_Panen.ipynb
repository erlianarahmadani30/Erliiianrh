```python
def hitung_panen_jagung():
    print("=========================================")
    print("   PROGRAM PENGHITUNG HASIL PANEN JAGUNG")
    print("=========================================")

    try:
        # Input data dari pengguna
        luas_lahan = float(input("Masukkan luas lahan (dalam hektar): "))
        hasil_per_hektar = float(input("Masukkan estimasi hasil per hektar (dalam kg): "))
        harga_per_kg = float(input("Masukkan harga jual per kg (Rp): "))

        # Proses Perhitungan
        total_berat_kg = luas_lahan * hasil_per_hektar
        total_berat_ton = total_berat_kg / 1000
        total_pendapatan = total_berat_kg * harga_per_kg

        # Menampilkan Hasil
        print("\n" + "="*40)
        print("          LAPORAN HASIL PANEN")
        print("="*40)
        print(f"Luas Lahan          : {luas_lahan} Hektar")
        print(f"Total Hasil Panen   : {total_berat_kg:,.2f} kg ({total_berat_ton:,.2f} Ton)")
        print(f"Estimasi Pendapatan : Rp {total_pendapatan:,.2f}")
        print("="*40)

    except ValueError:
        print("\n[Error] Input tidak valid! Harap masukkan angka yang benar.")

if __name__ == "__main__":
    hitung_panen_jagung()
```
